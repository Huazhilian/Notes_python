**Stack (Last-In-First-Out)**

````python
class Stack:
    def __init__(self):
        self.Stack=[]
    
    def push(self, item):
        self.Stack.append(item)
    
    def pop(self, index=-1):
        if not self.is_empty():
            return self.Stack.dequeue(index)
        else:
            raise Exception("Stack is empty")
        
    def peek(self, index=-1):
        if not self.is_empty():
            return self.Stack[index]
        else:
            raise Exception("Stack is empty")
        
    def is_empty(self):
        return len(self.Stack)==0
    
    def size(self):
        return len(self.Stack)
    
stack=Stack()
stack.enqueue(1);print(stack.Stack)  # enqueue
stack.enqueue(2);print(stack.Stack)  # enqueue
stack.enqueue(3);print(stack.Stack)  # enqueue
top=stack.pop();print(top,stack.Stack)  # dequeue
top2=stack.peek();print(top2,stack.Stack)  # peek
is_empty=stack.is_empty();print(is_empty)  # check if empty
size=stack.size();print(size)  # check size
````

**queue (First-In-First-Out)**

````python
class Queue:
    def __init__(self):
        self.q=[]
    
    def enqueue(self, item):
        self.q.append(item)
    
    def dequeue(self, index=0):
        if not self.is_empty():
            return self.q.pop(index)
        else:
            raise Exception("Queue is empty")
        
    def peek(self, index=0):
        if not self.is_empty():
            return self.q[index]
        else:
            raise Exception("Queue is empty")
        
    def is_empty(self):
        return len(self.q)==0
    
    def size(self):
        return len(self.q)
    
queue=Queue()
queue.enqueue(1);print(queue.q)  # push
queue.enqueue(2);print(queue.q)  # push
queue.enqueue(3);print(queue.q)  # push
top=queue.dequeue();print(top, queue.q)  # dequeue
top2=queue.peek();print(top2, queue.q)  # peek
is_empty=queue.is_empty();print(is_empty)  # check if empty
size=queue.size();print(size)  # check size
````

- ### Linked list/List node

**node: data (value) & pointer (next,[previous])**

**singly linked list: head & next**

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def reverse_pairs (self):
        if not self.head or not self.head.next:
            return
        previous = None
        current = self.head
        self.head = current.next
        
        while current and current.next:
            next_node = current.next
            next2_node = next_node.next

            # Swap current pair
            next_node.next = current
            if previous:
                previous.next = next_node

            # Prepare for next loop
            current.next = next2_node
            previous = current
            current = next2_node            

    def display(self):
        current = self.head
        while current.next:
            print(current.value, end='->')
            current = current.next
        print(current.value)

# Example usage of LinkedList
sll = SinglyLinkedList()
sll.append(1)
sll.append(2)
sll.append(3)
sll.append(4)
sll.append(5)
sll.append(6)
sll.display()  # Display the linked list
sll.reverse_pairs()  # Reverse pairs in the linked list
sll.display()  # Display the linked list after reversing pairs
````

**doubly linked list: head & next, previous**

````python
class DNode:
    def __init__ (self, value):
        self.value = value
        self.previous = None
        self.next = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, value):
        new_node = DNode(value)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node
        new_node.previous = current

    def display_forward(self):
        current = self.head
        while current.next:
            print(current.value, end=' <-> ')
            current = current.next
        print(current.value)

    def display_backward(self):
        current = self.head
        while current.next:
            current = current.next
        while current.previous:
            print(current.value, end=' <-> ')
            current = current.previous
        print(current.value)
    
# Example usage of DoublyLinkedList
dll = DoublyLinkedList()
dll.append(1)
dll.append(2)
dll.append(3)
dll.display_forward()  # Display the doubly linked list forward
dll.display_backward()  # Display the doubly linked list backward
````

**singly / doubly (circular) linked list (with tail)**

````python
class cNode:
    def __init__(self, value):
        self.value = value
        self.next = None
        self.previous = None # for doubly linked list

class CircularLinkedList:
    def __init__(self, use_circular=False, use_tail=False):
        self.use_circular = use_circular
        self.use_tail = use_tail
        self.head = None 
        self.tail = None if use_tail else None

    def append(self, value):
        new_node = cNode(value)
        if not self.head:
            self.head = new_node
            if self.use_tail:
                self.tail = new_node
            if self.use_circular:
                new_node.next = new_node
                new_node.previous = new_node
            return
        
        if self.use_tail:
            self.tail.next = new_node
            new_node.previous = self.tail
            self.tail = new_node
            if self.use_circular:
                new_node.next = self.head
                self.head.previous = new_node
    
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node
            new_node.previous = current
            if self.use_circular:
                new_node.next = self.head
                self.head.previous = new_node
    
    def display(self,limit=5):
        current = self.head
        try:        
            if self.use_circular:
                for i in range(limit-1):
                    print(current.value, end='<->')
                    current = current.next
                print(current.value, "(circular)")
            else:
                while current.next:
                    print(current.value, end='<->')
                    current = current.next
                print(current.value)
        except:
            print("Please add nodes into the linked list to display")
     
# Example usage of circular LinkedList
cll = CircularLinkedList(use_circular=True, use_tail=True)
cll.display(9)  # empty linked list raise error
cll.append(1)
cll.append(2)
cll.append(3)
cll.display(9)  # Display the circular linked list with limit 9

# Example usage of non-circular LinkedList
cll_non_circular = CircularLinkedList(use_circular=False, use_tail=False)
cll_non_circular.append(1)
cll_non_circular.append(2)
cll_non_circular.append(3)
cll_non_circular.display()  # Display the non-circular linked list 
````

- ### Binary (search) tree

**binary tree**

structure:
```code
         root
        /    \
left-child  right-child
```

traversals:

1. preorder (root - left - right)

2. inorder (left - root - right)

3. postorder (left - right - root)

4. level order

**binary search tree**

left-tree < root < right-tree

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        if not self.root:
            self.root = Node(value)
        else:
            self._insert_recursive(self.root, value)

    def _insert_recursive(self, node, value):
        if value < node.value:
            if node.left is None:
                node.left = Node(value)
            else:
                self._insert_recursive(node.left, value)
        else:
            if node.right is None:
                node.right = Node(value)
            else:
                self._insert_recursive(node.right, value)

    def inorder_traversal(self, node):
        if node:
            self.inorder_traversal(node.left)
            print(node.value, end=' ')
            self.inorder_traversal(node.right)

# Example usage of BinaryTree
bt = BinarySearchTree()
bt.insert(5)
bt.insert(3)
bt.insert(7)
bt.insert(2)
bt.insert(4)
bt.insert(6)
bt.insert(8)
print("Inorder Traversal of Binary Tree:")
bt.inorder_traversal(bt.root)  # Display the binary tree in inorder traversal
````
