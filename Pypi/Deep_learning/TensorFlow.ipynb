```python
import tensorflow as tf
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])
c = tf.matmul(a, b)
d = a + b
print(c.numpy())
print(d.numpy())
```

**gradient**

```python
import tensorflow as tf

x = tf.Variable([[1.0, 2.0], [3.0, 4.0]])
with tf.GradientTape() as tape:
    y = tf.exp(x)
grad = tape.gradient(y, x)
print(grad.numpy())
```
