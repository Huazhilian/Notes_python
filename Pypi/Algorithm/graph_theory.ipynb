- ### Path planning

Given graph $G=(V,E)$ and weights $w(u,v)\geq0$, find the shortest distance $d(t)$ from source s to target t.

**Dijkstra’s Algorithm**

Steps:

1. Initialize best-known distance d and knwon shortest distance set S

$$S=\{s\},$$

$$d(s)=0, \quad d(v)=\infty \text{ , for} \quad v\neq s.$$

2. While $S\neq V$:

$$S=S\bigcup\{u\},\quad u=arg\min_{v\notin S} d(v)$$

$$d(v)=\min\{d(v),d(u)+w(u,v)\}$$

```python
import math
def w(a,b):
    for u,v,w in edges:
        if u == a and v == b:
            return w
    return math.inf
        
def dijksta(n, edges, start):
    S=[]
    D=[math.inf]*n
    D[start]=0
    previous_node = [None]*n
    while len(S)<n:
        next_node=D.index(min(D[i] for i in range(n) if i not in S))
        S.append(next_node)
        for i in range(n):
            if i not in S:
                for j in S:
                    if (d:=D[j]+w(j,i)) < D[i]:
                        D[i]=d
                        previous_node[i]=j
    return D, previous_node

def reconstruct_path(pre, source, target):
    path = []
    while target is not None:
        path.append(target)
        target = pre[target]
    path.reverse()
    if source in path:
        return path[source:]
    else:
        return []  # No path found

edges = [(0,1,2),(0,2,4),(1,2,1),(1,3,7),(2,3,3)]
n = 4
source = 0
distances, pre = dijksta(n, edges, source)
print("Shortest distances from node", source, ":", distances)
print("Previous nodes in the shortest path tree:", pre)

source_node = 0
target_node = 3
path = reconstruct_path(pre, source_node, target_node)
print("Shortest path from node", source_node, "to node", target_node, ":", path)
```

**A\***

based on Dijksta algorithm, the cost function to be minimized in A* is 
$$f(u)=d(u)+h(u)$$
where $d(u)$ is the distance function used in Dijksta algorithm, and $h(u)$ is the heuristic function, a measure for the cost from node $u$ to the target $t$.

For nodes in coordinates, $h(u)$ can be choosen as some p-norm distance
$$h(u)=(\sum_i(x_{i,u}-x_{i,t})^p)^{\frac{1}{p}}.$$

For vertex in weighted graph, h(u)$ can be choosen as landmark heuristic, given the precomputed shortest paths from each landmark, $h(u)$ can be defined as
$$h(u)=\max_{L_i}|d(L_i,t)-d(L_i,u)|.$$

1. Initialize best-known distance d and knwon shortest distance set S

2. While $S\neq V$:

$$S=S\bigcup\{u\},\quad u=arg\min_{v\notin S} d(v)$$

$$d(v)=\min\{d(v),d(u)+w(u,v)\}$$
