**concat & merge**

- ### DataFrame

**construct from series**

````python
import pandas as pd

# create series
s1 = pd.Series([1, 2, 3, 4, 5]) # list-like object
s2 = pd.Series(['a', 'b', 'c', 'd', 'e'])

# create dataframe, series as columns
df = pd.DataFrame({'Numbers': s1, 'Letters': s2}) # dictionary of series

# create dataframe, series as rows
df2 = pd.DataFrame([s1,s2]) # list of series
print(df,"\n")
print(df2)
````

**update**

````python
import pandas as pd
data = [[1, 'a'], [2, 'b'], [3, 'c']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'], index=['row1', 'row2', 'row3'])
print(df,"\n")
df['Numbers'] =df['Numbers'].astype(float) # change data type of a column
df['Letters'] =[x.upper() for x in df['Letters']] # modify a column
print(df)
````

**attribution**

````python
import pandas as pd 
data = [[1, 'a'], [2, 'a'], [1, 'a']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'])
print(df)
print(df.size)
print(df.shape)
print(df.index)
print(df.values)
print(df.value_counts())
````

**slicing**

````python
import pandas as pd
data = [[1, 'a'], [2, 'b'], [3, 'c']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'],index=['row1', 'row2', 'row3'])
print(df,"\n")
print(df[1:3],"\n") # by position
print(df[1:3][1:2],"\n") # chaining: slicing [1:2] of df[1:3]
print(df['Numbers']) # by column name
````

````python
import pandas as pd
data = [[1, 'a'], [2, 'b'], [3, 'c']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'],index=['row1', 'row2', 'row3'])
print(df,"\n")
print(df.head(2))
print(df.tail(2))
````

````python
import pandas as pd
data = [[1, 'a'], [2, 'b'], [3, 'c']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'],index=['row1', 'row2', 'row3'])
print(df,"\n")
print(df.iloc[0:2,1:],"\n") # index location, end not included
print(df.loc['row1':'row2','Numbers']) # label location, end included
````

**concat & merge**

````python
import pandas as pd
data1 = [[1, 'a'], [2, 'b'], [3, 'c']]
data2 = [[1, 'd'], [2, 'e'], [6, 'f']]
df1 = pd.DataFrame(data1, columns=['Numbers', 'Letters1'],index=['row1', 'row2', 'row3'])
df2 = pd.DataFrame(data2, columns=['Numbers', 'Letters2'],index=['row1', 'row2', 'row3'])
print(df1,"\n")
print(df2,"\n")
df3 = pd.concat([df1, df2], axis=0,join="outer") # concatenate by rows
df4 = pd.concat([df1, df2], axis=1) # concatenate by columns
df5 = pd.merge(df1, df2, on='Numbers',how="outer") # merge on common column
print(df3,"\n")
print(df4,"\n")
print(df5)
````

**split**

````python
import pandas as pd
data = [[1, 'a'], [2, 'b'], [3, 'c'], [2, 'c'], [1, 'b']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'],index=['row1', 'row2', 'row3', 'row4', 'row5'])
print(df,"\n")
df_pivot = df.pivot(columns='Numbers', values='Letters')
df_melt = pd.melt(df, id_vars=['Numbers'], value_vars=['Letters'])
print(df_pivot,"\n")
print(df_melt)
````

**filter**

````python
import pandas as pd
data = [[1, 'a'], [2, 'b'], [3, 'c']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'],index=['row1', 'row2', 'row3'])
print(df,"\n")
df1 = df[df['Numbers']%3 == 0]
df2 = df[(df['Numbers'] >= 1) & (df['Numbers'] <= 2)]
print(df1,"\n")
print(df2)
````

**sort**

````python
import pandas as pd
data = [[3, 'c'], [1, 'b'], [2, 'a']]
df = pd.DataFrame(data, columns=['Numbers', 'Letters'],index=['row3', 'row1', 'row2'])
print(df,"\n")
print(df.sort_index(),"\n")
print(df.sort_values(by='Letters'))
````

**group by**

````python
import pandas as pd
data = {'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
        'age': [24, 26, 22, 24, 22],
        'salary': [50000, 60000, 55000, 70000, 65000]}
df = pd.DataFrame(data)
d1_g1 = df.groupby("age")["salary"].mean()
print(d1_g1, "\n")
d1_g2 = df.groupby("age").agg({'salary':['mean','max'],'name':'count'})
print(d1_g2)
````

**operation**

````python
import pandas as pd
data = [[i-j for i in range(3)] for j in range(4)]
df1 = pd.DataFrame(data)
df2 = pd.DataFrame(data)
print("df1\n",df1,"\n")
print("df2\n",df2,"\n")
df3 = df1 + df2
df4 = df1 * df2
df5 = df1 * 2 + df2
print("df3\n",df3,"\n")
print("df4\n",df4,"\n")
print("df5\n",df5)
````

**apply with lambda**

````python
import pandas as pd
data = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
df = pd.DataFrame(data)
df_apply = df.apply(lambda x: x*2)
df_apply2 = df[0].apply(lambda x: x+10)
print(df_apply, "\n")
print(df_apply2)
````
