- ### Create a new project
1. create a project with name *proj_name*
```
uv init *proj_name*
```
2. go inside the project
```
cd *proj_name*
```

---

- ### Create & active a virtual environment

1. 
```
uv venv myenv
```

2. 
```myenv/Scripts/activate```

---

- ### Add/remove dependencies
1. if has ```requirements.txt```

```
pip install -r requirements.txt
```

2. if cloned a repo with ```pyproject.toml``` 
```
uv sync
```

3. Update pyproject.toml

```
uv add *dependencies*
```
```
uv remove *dependencies*
```


---

- ### Run code
1. run python code
```
uv run *file_name.py*
```

2. run jupyter notebook code
```
uv run jupyter notebook
```
---

- ### Build project

1. build project
```
uv build
```
where the output goes to ```dist/```

---

- ### Publish to PyPI

1. 
```
uv publish
```

---

- ### Remove environment
1. 
```
rm -r .venv
```

---

- ### Overall
```
uv init myproject
cd myproject
uv venv
source .venv/bin/activate     # or Scripts/Activate.ps1 on Windows

uv add requests

uv run python src/myproject/main.py

uv build
uv publish
```
