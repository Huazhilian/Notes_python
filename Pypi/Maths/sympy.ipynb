- ### Expressions

**symbol**

```python
import sympy as sp
a = sp.Symbol('a')
b = sp.Symbol('b')
expr = sp.sin(a)**2 + sp.cos(b)**2
print(expr)
```

```python
import sympy as sp
x, y, z = sp.symbols('x y z')
expr = sp.exp(x) * sp.sin(y) + sp.log(z)
print(expr)
```

```python
import sympy as sp
l = 5
variables1 = sp.symbols('x1:3')
variables2 = sp.symbols('x1:' + str(l+1))
expr1 = sum(pow(variables1[i],2) for i in range(2))
expr2 = sum(pow(variables2[i],3) for i in range(l))
print(expr1)
print(expr2)
```

```python
import sympy as sp
x, y, z = sp.symbols('x y z')
expr = sp.sqrt(x**2 + y**2 + z**2)
print(expr)
```

**simplify**

```python
import sympy as sp
x, y = sp.symbols('x y')
expr = (x + 6) * 2 + (x + 6) * y
print(expr)
expr_simplified = expr.simplify()
print(expr_simplified)
```

```python
import sympy as sp
import numpy as np
p = sp.nsimplify(3.1415)
pi = sp.nsimplify(sp.pi)
a = sp.nsimplify(np.float64(-2.0))
print(p)
print(pi)
print(a)
```

**function**

```python
import sympy as sp
x, y = sp.symbols('x y')
F = sp.Function('F')(x, y)
expr = F + sp.diff(F, x) + sp.diff(F, y, 2)
print(expr)
```

**numerical evaluation**

```python
import sympy as sp
x, y = sp.symbols('x y')
expr = x**5 + 3*y**4 - 2*x**3 + y - 7
val = expr.evalf(subs={x: 2.5, y: -1.5})
val_high = expr.evalf(subs={x: 2.5, y: -1.5}, n=20)
print(val)
print(val_high)
```

**numerical/symbolic substitution**

```python
import sympy as sp
x = sp.Symbol('x')
y = sp.Symbol('y')
expr = sp.sin(x)
r1 = expr.subs(x, y + sp.pi/2)
r2 = expr.subs({x: y + sp.pi})
print(r1)
print(r2)
```

- ### Math function

**arithmetic**

```python
import sympy as sp
x, y = sp.symbols('x y')
expr1 = sp.ceiling(x) + sp.floor(y)
expr2 = sp.Abs(x - y)
expr3 = sp.sign(x - y)
expr4 = sp.root(x**3 - y**3, 3) + sp.sqrt(x + y)
expr5 = sp.factorial(sp.floor(x))
expr6 = sp.binomial(x, 2) + sp.binomial(y, 3)
print(expr1,"\n",expr2,"\n",expr3,"\n",expr4,"\n",expr5,"\n",expr6)
```

**piecewise function**

```python
import sympy as sp
x = sp.Symbol('x')
p = sp.Piecewise((x**2, (x < -8) | (x > 5)), (x + 1, (x >= -5) & (x <= 5)), (10, True))
print(p.evalf(subs={x: -6}), p.evalf(subs={x: 2}), p.evalf(subs={x: 10}))
sp.plot(p, (x, -10, 10), title='Piecewise Function', ylabel='p(x)', xlabel='x')
```

**Dirac delta function**

```python
import sympy as sp
x = sp.Symbol('x')
delta = sp.DiracDelta(x - 2)
print(delta.evalf(subs={x: 2}), delta.evalf(subs={x: 3}))
```

**unit Step function**

```python
import sympy as sp
x = sp.Symbol('x')
UnitStep = sp.Heaviside(x - 2)
print(UnitStep.evalf(subs={x: 0}),UnitStep.evalf(subs={x: 2}), UnitStep.evalf(subs={x: 3}))
sp.plot(UnitStep, (x, 0, 5), title='Unit Step Function', ylabel='H(x-2)', xlabel='x')
```

**exp/log**

```python
import sympy as sp
x, y = sp.symbols('x y')
expr = sp.log(x**2 + y**2) + sp.exp(x*y)
print(expr)
```

**triangular function**

```python
import sympy as sp
l = 4
variables = sp.symbols('x1:' + str(l+1))
expr1 = sp.sin(variables[0]) + sp.cos(variables[0]) + sp.tan(variables[0])
expr2 = sp.sec(variables[0]) + sp.csc(variables[0]) + sp.cot(variables[0])
expr3 = sp.asin(variables[0]) + sp.acos(variables[0]) + sp.atan(variables[0])
expr4 = sp.asec(variables[0]) + sp.acsc(variables[0]) + sp.acot(variables[0])
expr5 = sp.sinh(variables[0]) + sp.cosh(variables[0]) + sp.tanh(variables[0])
expr6 = sp.sech(variables[0]) + sp.csch(variables[0]) + sp.coth(variables[0])
expr7 = sp.asinh(variables[0]) + sp.acosh(variables[0]) + sp.atanh(variables[0])
expr8 = sp.asech(variables[0]) + sp.acsch(variables[0]) + sp.acoth(variables[0])
print(expr1,"\n",expr2,"\n",expr3,"\n",expr4,"\n",expr5,"\n",expr6,"\n",expr7,"\n",expr8)
```

**complex**

```python
import sympy as sp
x, y, z= sp.symbols('x y z')
z = x + sp.I*y
z = z.evalf(subs={x: 3, y: 4})
n = sp.re(z) + sp.im(z)
arg = sp.arg(z)
abs = sp.Abs(z)
conj = sp.conjugate(z)
print(n,"\n",arg,"\n",abs,"\n",conj)
```

**special funtion**

```python
import sympy as sp
x, n = sp.symbols('x n')
gamma = sp.gamma(x)
beta = sp.beta(x, x + 1)
polygamma = sp.polygamma(n, x)
print(gamma.evalf(subs={x: 5.5}))
print(beta.evalf(subs={x: 2.5}))
print(polygamma.evalf(subs={n: 1, x: 3.0}))
```

```python
import sympy as sp
x, n = sp.symbols('x n')
besselj = sp.besselj(n, x)
bessely = sp.bessely(n, x)
print(besselj.evalf(subs={n: 2, x: 3.0}))
print(bessely.evalf(subs={n: 2, x: 3.0}))
```

```python
import sympy as sp
x, n = sp.symbols('x n')
legendre = sp.legendre(n, x)
hermite = sp.hermite(n, x)
chebyshevT = sp.chebyshevt(n, x)
chebyshevU = sp.chebyshevu(n, x)
print(legendre.evalf(subs={n: 3, x: 0.5}))
print(hermite.evalf(subs={n: 2, x: 1.0}))
print(chebyshevT.evalf(subs={n: 4, x: 0.5}))
print(chebyshevU.evalf(subs={n: 4, x: 0.5}))
```

```python
import sympy as sp
x= sp.symbols('x')
erf = sp.erf(x)
erfc = sp.erfc(x)
print(erf.evalf(subs={x: 1.0}))
print(erfc.evalf(subs={x: 1.0}))
```

- ### Calculus

**limits**

```python
import sympy as sp
x = sp.symbols('x')
f = sp.sin(x)/x
lim = sp.limit(f, x, 0)
print(lim)
```

```python
import sympy as sp
x = sp.Symbol('x')
p = sp.Piecewise((-1, x < 0), (0, x == 0), (1, x > 0))
limit_left = sp.limit(p, x, 0, dir='-')
limit_right = sp.limit(p, x, 0, dir='+')
print(limit_left, limit_right)
sp.plot(p, (x, -10, 10), title='Piecewise Function', ylabel='p(x)', xlabel='x')
```

**derivative**

```python
import sympy as sp
x, y = sp.symbols('x y')
F = sp.Function('F')(x, y)

# General derivative of F with respect to x and y
dF_dx = sp.diff(F, x)
dF_dy = sp.diff(F, y)

print("Derivative of F with respect to x:", dF_dx)
print("Derivative of F with respect to y:", dF_dy)
```

```python
import sympy as sp
x, y = sp.symbols('x y')
f = sp.sin(x) * sp.exp(y) + sp.cos(y) * sp.exp(x)
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

# Second-order partial derivatives
d2f_dx2 = sp.diff(f, x, 2)
d2f_dy2 = sp.diff(f, y, 2)
d2f_dxdy = sp.diff(df_dx, y)
print("First derivative with respect to x:", df_dx)
print("First derivative with respect to y:", df_dy)
print("Second derivative with respect to x:", d2f_dx2)
print("Second derivative with respect to y:", d2f_dy2)
print("Mixed second derivative:", d2f_dxdy)
```

```python
import sympy as sp
x, y = sp.symbols('x y')

# derivatives of an implicit function
F = x**2 + y**2 - 1
dy_dx = -sp.diff(F, x) / sp.diff(F, y)
 
print("\nDerivative of F(x,y)=0:\ndy/dx =", dy_dx)
```

**integrals**

```python
import sympy as sp
x = sp.symbols('x')
f = sp.sin(x) 
f_int = sp.integrate(f, x)
print(f_int)
```

```python
import sympy as sp
x = sp.symbols('x')
f = sp.sin(x) * sp.exp(x)
f_int_interval = sp.integrate(f, (x, 0, sp.pi))
print(f_int_interval)
```

```python
import sympy as sp
x, y = sp.symbols('x y')
f = sp.sin(x) * sp.exp(y) + sp.cos(y) * sp.exp(x)
f_int_interval = sp.integrate(f, (x, 0, sp.pi), (y, 0, sp.pi))
print(f_int_interval)
```

**series**

```python
import sympy as sp
x= sp.symbols('x')
f = sp.exp(x)
f_series_expansion = sp.series(f, x, 0, 6)
print(f_series_expansion)
```

```python
import sympy as sp
x, y = sp.symbols('x y')
f = sp.exp(x * y + x + y)
f_series_expansion_x = sp.series(f, x, 0, 4).removeO()
f_series_expansion = sp.series(f_series_expansion_x, y, 0, 4).removeO()
print(f_series_expansion.simplify())
```

- ### Linear Algebra

```python
import sympy as sp
u = sp.Matrix([1, 2, 3, 4])
v = sp.Matrix([5, 6, 7, 8])
c = u + v
d = u.T @ v
print("c =", c)
print("d =", d)
```

```python
import sympy as sp
a = sp.Matrix([[1, 2], [3, 4]])
b = sp.Matrix([[5, 6], [7, 8]])
c = a + b
d = a * b
print("Matrix Addition:\n", c)
print("Matrix Multiplication:\n", d)
```

```python
import sympy as sp
a = sp.Matrix([[1, 2], [3, 4]])
det = a.det()
trace = a.trace()
rank = a.rank()
inv = a.inv()
print("Determinant:", det)
print("Trace:", trace)
print("Rank:", rank)
print("Inverse:\n", inv)
```

```python
import sympy as sp
a = sp.Matrix([[1, 2], [3, 4]])
b = sp.Matrix([[5], [6]])
rref, sol = a.rref(b) # reduced row echelon form, solution to Ax=b
eigenvals = a.eigenvals()
eigenvects = a.eigenvects()
print("RREF:\n", rref)
print("\nSolution:\n", sol)
print("\nEigenvalues:", eigenvals)
for eigenv in eigenvects:
    print("\nEigenvector:", eigenv)
```
