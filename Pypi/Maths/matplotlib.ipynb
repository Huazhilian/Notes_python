**plot curves in one section**

```python
import numpy as np
from matplotlib import pyplot as plt
from math import cos,sin,pi

x=np.linspace(-pi,pi,100)
y=list(map(sin,x))
z=list(map(cos,x))

plt.title('cos(x) & sin(x)')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.plot(x,y,'g',label='sin(x)')
plt.plot(x,z,'y',label='cos(x)')
plt.legend()
plt.show()
```

**plot curves in subsections**


```python
import numpy as np
from matplotlib import pyplot as plt
from math import exp

x=np.linspace(-1,1,100)
y=list(map(exp,x))
z=x**2+1

plt.subplot(2,1,1)
plt.plot(x,y,'y',label='exp(x)')
plt.title('y=exp(x)')
plt.legend()

plt.subplot(2,1,2)
plt.plot(x,z,'c',label='x^2+1')
plt.title('y=x^2+1')
plt.legend()
plt.show()
```

- ### Bar chart

**plot bar chart**


```python
import numpy as np
from matplotlib import pyplot as plt

x1=np.arange(1,13,2)
y1=np.array([15,9,2,4,6,8])
x2=np.arange(2,13,2)
y2=np.array([5,3,9,12,7,6])

plt.bar(x1,y1,align='center')
plt.bar(x2,y2,align='center')
plt.title('Bar chart')
plt.xlabel('Month')
plt.ylabel('Volume')
plt.show()
```

- ### Multiple subplots

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate some sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)
y2 = np.cos(x)
np.random.seed(0)
rand_x = np.random.rand(50)
rand_y = np.random.rand(50)

# Create a figure with multiple subplots
fig, axs = plt.subplots(4, 3, figsize=(14, 12))
axs = axs.ravel() # Flatten the 2D array of axes for easy iteration

# 1. Line plot
axs[0].plot(x, y, label='sin(x)')
axs[0].plot(x, y2, label='cos(x)')
axs[0].set_title("Line Plot")
axs[0].legend()

# 2. Scatter plot
axs[1].scatter(rand_x, rand_y, c='red')
axs[1].set_title("Scatter Plot")

# 3. Bar chart
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]
axs[2].bar(categories, values)
axs[2].set_title("Bar Chart")

# 4. Horizontal bar chart
axs[3].barh(categories, values)
axs[3].set_title("Horizontal Bar Chart")

# 5. Histogram
data = np.random.randn(1000)
axs[4].hist(data, bins=30, color='orange', edgecolor='black')
axs[4].set_title("Histogram")

# 6. Box plot
axs[5].boxplot([data, np.random.randn(1000)])
axs[5].set_title("Box Plot")

# 7. Stack plot
axs[6].stackplot(x, np.abs(y), np.abs(y2), labels=['|sin(x)|', '|cos(x)|'])
axs[6].legend()
axs[6].set_title("Stack Plot")

# 8. Area plot (filled line plot)
axs[7].fill_between(x, y, alpha=0.3)
axs[7].set_title("Area Plot")

# 9. Pie chart
axs[8].pie(values, labels=categories, autopct='%1.1f%%', startangle=90)
axs[8].set_title("Pie Chart")

# 10. Stem plot
axs[9].stem(x[:10], y[:10])
axs[9].set_title("Stem Plot")

# 11. Step plot
axs[10].step(x, y)
axs[10].set_title("Step Plot")

# 12. Polar plot
axs[11] = plt.subplot(4, 3, 12, polar=True)
theta = np.linspace(0, 2*np.pi, 100)
axs[11].plot(theta, np.sin(5*theta))
axs[11].set_title("Polar Plot", va='bottom')

plt.tight_layout()
plt.show()
```

- ### 3D plot

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Generate sample data for 3D plots
X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create figure with multiple plots
fig = plt.figure(figsize=(16, 4))

# 3D Surface plot
ax1 = fig.add_subplot(1, 3, 1, projection='3d')
ax1.plot_surface(X, Y, Z, cmap='viridis')
ax1.set_title("3D Surface Plot")

# 3D Wireframe plot
ax2 = fig.add_subplot(1, 3, 2, projection='3d')
ax2.plot_wireframe(X, Y, Z, color='black')
ax2.set_title("3D Wireframe Plot")

# 16. Contour plot
ax3 = fig.add_subplot(1, 3, 3)
contour = ax3.contourf(X, Y, Z)
fig.colorbar(contour, ax=ax3)
ax3.set_title("Contour Plot")

plt.tight_layout()
plt.show()
```
