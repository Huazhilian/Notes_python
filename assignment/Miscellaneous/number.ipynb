```python
def compress(s):
    l=list(s)
    i=0
    n=len(l)
    while i < n:
        if l[i].isdigit():
            l.pop(i)
            n -= 1
        else:
            i += 1
    return ''.join(l)

s="abc1234d"
print(compress(s))
```

```python
def iterate(n):
    s = 0
    if n<10:
        return n
    while n>=10:
        s += n % 10
        n = n // 10
    s += n
    return iterate(s)

n=12345
print(iterate(n))
```

```python
class zeroddeven():
    def __init__(self, n):
        self.n = n
    
    def zero(self,i):
        if i%2 == 0:
            print(0, end='')
    
    def odd(self,i):
        if i%4 == 1:
            print((i+1)//2, end='')

    def even(self,i):
        if i%4 == 3:
            print((i+2)//2, end='')

zoe = zeroddeven(5)
for i in range(zoe.n*2):
    zoe.zero(i)
    zoe.odd(i)
    zoe.even(i)
```

```python
import numpy as np
def int2bin(i,k):
    a = np.array(list(map(int,list(bin(i)[2:]))))
    if len(a) < k:
        a = np.concatenate((np.zeros(k-len(a),dtype=int),a))
    return a

def halfable(nums):
    s = sum(nums)
    if s%2 == 1:
        return False
    print("half sum:",s//2)
    k = 0
    for j in range(2**len(nums)):
        if int2bin(j,len(nums)) @ nums == s//2: # @ is dot product
            k = 1
            p = int2bin(j,len(nums))*nums
            p1 = p[p != 0]
            q = nums-p
            p2 = q[q != 0]
            print(p1,p2)

    if k==1:
        return True
halfable([1,5,11,5])
```
