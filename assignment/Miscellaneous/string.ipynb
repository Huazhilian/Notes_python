```python
#matching game: iterable removes adjacent matching character
a ='cbbqqcde'
def e(x):
    if len(x)==0:
            return 0
    if len(x)==1:
            return x
    for i in range(1,len(x)):
        if x[i]==x[i-1]:
            x=x[:i-1]+x[i+1:]
            return e(x)
    return x
    
print(e(a))
```

```python
#matching game: removes adjacent matching character
a ='cbbqqcde'
def e(x):
    if len(x)==0:
            return 0
    if len(x)==1:
            return x
    
    i = 1  # Start from the second character
    while i < len(x):
        if x[i]==x[i-1]:
            x=x[:i-1]+x[i+1:]
            i-=1  # Adjust index after removal
        i += 1
    return x
    
print(e(a))
```

```python
class solution:
    def longest_common_subseq(self,str1, str2):
        m = len(str1)
        n = len(str2)
        d={}
        for i in range(m):
            r=""
            j=0
            while (j < n):
                if (str1[i] == str2[j]):
                    k=0
                    while (i+k< m and j+k < n and str1[i+k] == str2[j+k]):
                        r += str1[i+k]
                        k += 1
                j += 1
            d[i]=r
        print(d)
        return [value for value in d.values() if len(value) == max(len(v) for v in d.values())]
    
s=solution()
print(s.longest_common_subseq("abcdefg","abcefg"))
```
