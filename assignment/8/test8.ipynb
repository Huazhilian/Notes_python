```python
import numpy as np
def int2bin(i,k):
    a = np.array(list(map(int,list(bin(i)[2:]))))
    if len(a) < k:
        a = np.concatenate((np.zeros(k-len(a),dtype=int),a))
    return a

def halfable(nums):
    s = sum(nums)
    if s%2 == 1:
        return False
    print("half sum:",s//2)
    k = 0
    for j in range(2**len(nums)):
        if int2bin(j,len(nums)) @ nums == s//2: # @ is dot product
            k = 1
            p = int2bin(j,len(nums))*nums
            p1 = p[p != 0]
            q = nums-p
            p2 = q[q != 0]
            print(p1,p2)

    if k==1:
        return True
halfable([1,5,11,5])
```
