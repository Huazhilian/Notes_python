```python
def compress(s):
    l=list(s)
    i=0
    n=len(l)
    while i < n:
        if l[i].isdigit():
            l.pop(i)
            n -= 1
        else:
            i += 1
    return ''.join(l)

s="abc1234d"
print(compress(s))
```

```python
def iterate(n):
    s = 0
    if n<10:
        return n
    while n>=10:
        s += n % 10
        n = n // 10
    s += n
    return iterate(s)

n=12345
print(iterate(n))
```

```python
class zeroddeven():
    def __init__(self, n):
        self.n = n
    
    def zero(self,i):
        if i%2 == 0:
            print(0, end='')
    
    def odd(self,i):
        if i%4 == 1:
            print((i+1)//2, end='')

    def even(self,i):
        if i%4 == 3:
            print((i+2)//2, end='')

zoe = zeroddeven(5)
for i in range(zoe.n*2):
    zoe.zero(i)
    zoe.odd(i)
    zoe.even(i)
```
