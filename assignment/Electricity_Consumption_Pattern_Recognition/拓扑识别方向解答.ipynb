- ### 拓扑识别方向

1. 
输入：
所有用户A相用电数据，即包括电流，电压采样数据
$X = \{\bold{X}_i(t)\}_{i=1,\dots,N,t=t_1,\dots,t_T}=\{x_i(t), y_i(t)\}_{i=1,\dots,N,t=t_1,\dots,t_T}$ 其中$x_i(t)$为瞬时电流，$y_i(t)$为瞬时电压，$N$为采样点位数，$T$为单个采样点在采样频率为每15分钟一次的情况下总采样次数。

输出：
用户聚类结果标签$c_i\in\{1,\dots,K\}$，即在同一类代表用户在同一小区，假设共有$K$个小区即有$K$个类。

模型与方法：
首先进行数据预处理，清除无效值，数据单位标准化。
然后可使用Kmeans聚类方法，即对损失函数各数据点到簇中心距离平方$\sum_{j=1}^T\sum_{i=1}^k\sum_{x}\|\bold{X}_i(t_j)-\mu_i\|^2$进行最小函数值优化求解，其中$C_i$是第$i$个簇，$\mu_i$是第$i$个簇中心。

2. 
输入：
所有用户电流，电压采样数据同1，以及每个小区全部用户的总用电数据
$M = \{\bold{M}_i(t)\}_{i=1,\dots,N,t=t_1,\dots,t_T}=\{u_i(t), v_i(t)\}_{i=1,\dots,K,t=t_1,\dots,t_T}$ 其中$u_i(t)$为瞬时电流，$v_i(t)$为瞬时电压，$K$为小区数，$T$为单个采样点在采样频率为每15分钟一次的情况下总采样次数。

输出：
用户所属小区矩阵$A\in\{0,1\}^{\{K,N\}}$，其中$A_{ij}=1$代表用户$j$属于第$i$个小区。

模型与方法：
使用带约束条件的损失函数进行最小值函数优化求解，其中损失函数为各小区总用电到所有被指定为该小区的用户用电之和的距离平方的总计$\sum_{h=1}^T\sum_{i=1}^K\|\bold{M}_i(t_h)-\sum_{j=1}^NA_{ij}\bold{X}_j(t_h)\|^2$，其中约束条件为
$\sum_{i=1}^KA_{ij}=1,\forall j=1\dots N$。


3. 
影响：
如果每个表的采样时刻并不能完全对齐，可能会导致算法1中数据点的相关性降低，影响聚类效果，同样也可能导致算法2中各用户用电总和与总电表匹配度降低。

优化方法：
可考虑对数据在时间维度插值加密以获得公共时间戳上各用户的用电以及各小区的总用电，以尽可能获得同一时刻下各用户的用电数据及各小区的总用电数据，降低不同步的时间差带来的影响；同时也可考虑使用固定周期时间用电的平均值进行计算，以此消除瞬时的数据不对齐情况。

4. 
对于用户入住率很低的小区，其整体用电量低，数据变化不明显，可能会导致各用户用电量的差异性降低，聚类精确度降低，此时可考虑使用固定周期时间用电的平均值进行计算，以放大一些较小的周期性用电规律；同时也可考虑使用主成分分析，降低噪声对数据的影响，捕捉用电波动的特征。

5. 
- 尽可能获取真实有效的数据集，减少数据集中缺省值和异常值的存在。

- 对于各用户的用电数据，可同时采集其所在小区的标注代码，以获得更完整的数据。

- 尽可能保证各用户及总量数据采样的时间戳对齐，以降低对后期数据相关性的影响。

- 尽可能获取多维度的数据，可以采集用户的瞬时电流，瞬时电压，累计用电量，甚至是电表内部温度等，增加数据整体可靠度。

- 对于用户打开或关闭大功率电器等情形，可通过电表捕捉其所引起的瞬时电流，瞬时电压显著变化，一同采集到数据集中，增加数据特征。
