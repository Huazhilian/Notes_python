1. Djikstra algorithm

```python
import math
def w(a ,b, edges):
    for u,v,w in edges:
        if u == a and v == b:
            return w
    return math.inf
        
def dijksta(n, edges, start):
    S=[]
    D=[math.inf]*n
    D[start]=0
    while len(S)<n:
        next_node=D.index(min(D[i] for i in range(n) if i not in S))
        S.append(next_node)
        D = [D[i] if i in S else min(D[i], D[next_node] + w(next_node, i, edges)) for i in range(n)]
    return D

edges = [(0,1,2),(0,2,4),(1,2,1),(1,3,7),(2,3,3)]
n = 4
start = 0
distances = dijksta(n, edges, start)
print("Shortest distances from node", start, ":", distances)
```
