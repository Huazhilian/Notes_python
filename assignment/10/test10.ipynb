**badmintion score system**

```python
# Suppose two teams A and B are playing a badminton game with players A1, A2 and B1, B2 respectively.
class badminton_score:
    def __init__(self):
        self.team = [["A1", "A2"], ["B1", "B2"]] # In badminton court, players position in shape [[B1, B2],[A2, A1]]
        self.scores = [0, 0]
        self.swap = [0, 0]
        self.baseline = [3, 5]
        self.last_point = 0
        self.current_point = -1
        self.pos = ["right", "left"]

    def initial(self):
        self.get_scores()
        self.serves(0)

    def winner(self, Team):
        self.scores[Team] += 1
        self.current_point = Team
        self.serves(self.get_scores())
        self.last_point = Team

    def get_scores(self):
        if self.scores[0] >= self.baseline[0] or self.scores[1] >= self.baseline[0]:
            if abs(self.scores[0] - self.scores[1]) >= 2:
                if self.scores[0] > self.scores[1]:
                    print(f"score:{self.scores[0]}-{self.scores[1]}, Team 1 wins")
                    return 1
                else:
                    print(f"score:{self.scores[0]}-{self.scores[1]}, Team 2 wins")
                    return 1
        
            if self.scores[0] == self.baseline[1]:
                print("score:"f"{self.scores[0]}-{self.scores[1]}, Team 1 wins")
                return 1
            if self.scores[1] == self.baseline[1]:
                print("score:",f"{self.scores[0]}-{self.scores[1]}, Team 2 wins")
                return 1
        print("score:",f"{self.scores[0]}-{self.scores[1]}") 
        return 0

    def serves(self, r):
        if r == 0:
            if self.current_point == -1:
                self.current_point = 0
                print(f"Player A1 in Team 1 serves on the right")
                return
            if self.current_point == self.last_point:
                self.swap[self.current_point] = 1 - self.swap[self.current_point]
                    
            player = self.team[self.current_point][abs(self.scores[self.current_point]%2 - self.swap[self.current_point])]
            print(f"Player {player} in Team {self.current_point + 1} serves on the {self.pos[self.scores[self.current_point]%2]} ")
        
bs = badminton_score()
bs.initial()
bs.winner(0)
bs.winner(1)
bs.winner(0)
bs.winner(1)
bs.winner(0)
bs.winner(1)
bs.winner(0)
bs.winner(1)
bs.winner(0)
```
